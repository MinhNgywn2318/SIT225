```python
import random
import matplotlib.pyplot as plt

n_values = 100
y_values = []

# Create data (y_values) randomly between 1 and 100.
for i in range(n_values):
    y_values.append(random.randint(1, 100))

x_values = range(n_values)  # X is sequence of values 0-99
plt.plot(x_values, y_values)
plt.show()
```

```python
# Plot 2 variables
# 

n_values = 100
y_values_1 = []
y_values_2 = []

# Create data (y_values) randomly between 1 and 100.
for i in range(n_values):
    y_values_1.append(random.randint(1, 100))
    y_values_2.append(random.randint(1, 100))

x_values = range(n_values)  # X is sequence of values 0-99
plt.plot(x_values, y_values_1)
plt.plot(x_values, y_values_2)  # call plot again draws in the same graph.
plt.show()
```

```python
import matplotlib.pyplot as plt

# x values: sample index
x_values = list(range(0, 20))

# y values: increasing at a steady rate (here, rate = 3 per step)
y_values = [i * 3 for i in x_values]

plt.plot(x_values, y_values, marker='o')
plt.title("Activity 1: Ascending Line")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# x values: sample index
x_values = list(range(0, 20))

# y values: decreasing at a steady rate (here, rate = -4 per step), starting from a high value
y_values = [100 - i * 4 for i in x_values]

plt.plot(x_values, y_values, marker='o', color='orange')
plt.title("Activity 2: Descending Line")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import math

# x values: sample index (finer step for a smooth wave)
x_values = [i * 0.1 for i in range(0, 100)]

# y values: sine wave
y_values = [math.sin(x) for x in x_values]

plt.plot(x_values, y_values, color='green')
plt.title("Activity 3: Wave (sine)")
plt.xlabel("x")
plt.ylabel("y = sin(x)")
plt.grid(True)
plt.show()
```
